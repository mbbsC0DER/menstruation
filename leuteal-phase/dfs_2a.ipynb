# Practical 2a: Depth First Search

Run the cells in order. Code from `dfs_2a.py`.

```python
graph = {
    'A' : ['B', 'C'],
    'B' : ['D','E'],
    'C' : ['F','G'],
    'D' : [],
    'E' : [],
    'F' : [],
    'G' : []
}

visited = set()

def dfs(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbour in graph[node]:
            dfs(neighbour)

print('Travesal')
dfs('A')
```
